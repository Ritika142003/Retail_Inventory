```python
import duckdb

duckdb.sql("""
CREATE OR REPLACE TABLE retail_store_inventory AS
SELECT *
FROM read_csv_auto('retail_store_inventory.csv');
""")
```

```python
duckdb.sql("""
SELECT COUNT(*) AS Total_Rows
FROM retail_store_inventory;
""")
```

```python
duckdb.sql("""
SELECT
    SUM(Units_Sold) AS Total_Units_Sold,
    SUM(Inventory_Level) AS Total_Inventory,
    AVG(Units_Sold) AS Avg_Units_Sold,
    AVG(Inventory_Level) AS Avg_Inventory,
    SUM(Units_Ordered) AS Total_Units_Ordered,
    AVG(Price) AS Avg_Price,
    AVG(Discount) AS Avg_Discount
FROM retail_store_inventory;
""")
```

```python
duckdb.sql("""
SELECT
    SUM("Units Sold") AS Total_Units_Sold,
    SUM("Inventory Level") AS Total_Inventory,
    AVG("Units Sold") AS Avg_Units_Sold,
    AVG("Inventory Level") AS Avg_Inventory,
    SUM("Units Ordered") AS Total_Units_Ordered,
    AVG(Price) AS Avg_Price,
    AVG(Discount) AS Avg_Discount
FROM retail_store_inventory;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    SUM("Units Sold") AS Total_Units_Sold,
    SUM("Inventory Level") AS Total_Inventory,
    AVG("Inventory Level") AS Avg_Inventory,
    SUM("Units Ordered") AS Total_Units_Ordered
FROM retail_store_inventory
GROUP BY "Store ID"
ORDER BY Total_Units_Sold DESC;
""")
```

```python
duckdb.sql("""
SELECT
    Category,
    SUM("Units Sold") AS Total_Units_Sold,
    AVG("Units Sold") AS Avg_Units_Sold,
    SUM("Inventory Level") AS Total_Inventory,
    AVG("Inventory Level") AS Avg_Inventory
FROM retail_store_inventory
GROUP BY Category
ORDER BY Total_Units_Sold DESC;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    "Product ID",
    Date,
    "Inventory Level",
    "Units Sold"
FROM retail_store_inventory
WHERE "Inventory Level" < 100
ORDER BY "Inventory Level" ASC;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    "Product ID",
    Date,
    "Inventory Level",
    "Demand Forecast",
    "Units Sold",
    "Demand Forecast" - "Inventory Level" AS Shortage_Gap
FROM retail_store_inventory
WHERE "Demand Forecast" > "Inventory Level"
ORDER BY Shortage_Gap DESC;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    "Product ID",
    Date,
    "Inventory Level",
    "Demand Forecast",
    "Inventory Level" - "Demand Forecast" AS Inventory_Gap
FROM retail_store_inventory
ORDER BY Inventory_Gap ASC;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    "Product ID",
    Date,
    "Units Sold",
    "Inventory Level"
FROM retail_store_inventory
WHERE "Units Sold" > 300
  AND "Inventory Level" < 100
ORDER BY "Units Sold" DESC;
""")
```

```python
duckdb.sql("""
SELECT
    "Store ID",
    "Product ID",
    Date,
    "Inventory Level",
    "Demand Forecast",
    "Units Sold"
FROM retail_store_inventory
WHERE "Inventory Level" > 400
  AND "Demand Forecast" < 100
ORDER BY "Inventory Level" DESC;
""")
```

```python
duckdb.sql("""
SELECT
    "Product ID",
    SUM("Units Sold") AS Total_Units_Sold
FROM retail_store_inventory
GROUP BY "Product ID"
ORDER BY Total_Units_Sold DESC
LIMIT 5;
""")
```

```python
duckdb.sql("""
SELECT
    "Product ID",
    SUM("Units Sold") AS Total_Units_Sold
FROM retail_store_inventory
GROUP BY "Product ID"
ORDER BY Total_Units_Sold ASC
LIMIT 5;
""")
```

```python
duckdb.sql("""
SELECT
    COUNT(*) AS Total_Records,

    SUM(CASE
        WHEN "Inventory Level" < 100 THEN 1
        ELSE 0
    END) AS Low_Inventory_Records,

    SUM(CASE
        WHEN "Demand Forecast" > "Inventory Level" THEN 1
        ELSE 0
    END) AS Potential_Shortage_Records,

    SUM(CASE
        WHEN "Inventory Level" > 400
         AND "Demand Forecast" < 100 THEN 1
        ELSE 0
    END) AS Potential_Overstock_Records,

    SUM(CASE
        WHEN "Units Sold" > 300
         AND "Inventory Level" < 100 THEN 1
        ELSE 0
    END) AS High_Sales_Low_Inventory_Records

FROM retail_store_inventory;
""")
```

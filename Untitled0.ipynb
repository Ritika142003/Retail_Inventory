```python
import pandas as pd

df_clean = pd.read_csv('/content/retail_inventory_cleaned.csv')
```

```python
df_clean['Date'] = pd.to_datetime(df_clean['Date'])
```

```python
numeric_cols = [
    'Inventory Level',
    'Units Sold',
    'Units Ordered',
    'Demand Forecast',
    'Price',
    'Discount',
    'Competitor Pricing'
]

df_clean[numeric_cols].describe().T
```

```python

df_clean[numeric_cols].skew()
```

```python
categorical_cols = [
    'Store ID',
    'Category',
    'Region',
    'Weather Condition',
    'Seasonality'
]

for col in categorical_cols:
    print(f"\n{col}")
    print(df_clean[col].value_counts())
```

```python
df_clean.groupby('Region')['Units Sold'].sum().sort_values(ascending=False)
```

```python
sales_summary = pd.DataFrame({
    'By Category': df_clean.groupby('Category')['Units Sold'].sum(),
    'By Store': df_clean.groupby('Store ID')['Units Sold'].sum(),
    'By Region': df_clean.groupby('Region')['Units Sold'].sum()
})

sales_summary
```

```python
print("INVENTORY BY CATEGORY")
print(df_clean.groupby('Category')['Inventory Level'].mean().sort_values(ascending=False))

print("\nINVENTORY BY STORE")
print(df_clean.groupby('Store ID')['Inventory Level'].mean().sort_values(ascending=False))
```

```python
# STEP 5 — INVENTORY ANALYSIS

print("=== AVERAGE INVENTORY BY CATEGORY ===")
print(
    df_clean.groupby('Category')['Inventory Level']
    .mean()
    .sort_values(ascending=False)
)

print("\n=== AVERAGE INVENTORY BY STORE ===")
print(
    df_clean.groupby('Store ID')['Inventory Level']
    .mean()
    .sort_values(ascending=False)
)

print("\n=== OVERALL INVENTORY ===")
print("Average Inventory:", df_clean['Inventory Level'].mean())
print("Minimum Inventory:", df_clean['Inventory Level'].min())
print("Maximum Inventory:", df_clean['Inventory Level'].max())
```

```python
# STEP 6 — ACTUAL SALES VS DEMAND FORECAST

print("=== AVERAGE ===")
print("Average Actual Sales:", df_clean['Units Sold'].mean())
print("Average Demand Forecast:", df_clean['Demand Forecast'].mean())

print("\n=== TOTAL ===")
print("Total Actual Sales:", df_clean['Units Sold'].sum())
print("Total Demand Forecast:", df_clean['Demand Forecast'].sum())

print("\n=== DIFFERENCE ===")
difference = df_clean['Demand Forecast'].sum() - df_clean['Units Sold'].sum()
print("Forecast - Actual:", difference)
```

```python
# STEP 7 — TIME / SEASON ANALYSIS

print("=== SALES BY SEASON ===")
print(
    df_clean.groupby('Seasonality')['Units Sold']
    .mean()
    .sort_values(ascending=False)
)

print("\n=== SALES BY MONTH ===")
df_clean['Month'] = df_clean['Date'].dt.month

print(
    df_clean.groupby('Month')['Units Sold']
    .mean()
    .sort_values(ascending=False)
)

print("\n=== SALES BY YEAR ===")
df_clean['Year'] = df_clean['Date'].dt.year

print(
    df_clean.groupby('Year')['Units Sold']
    .mean()
)
```

```python
# STEP 8 — PROMOTION / DISCOUNT ANALYSIS

print("=== SALES BY PROMOTION ===")
print(
    df_clean.groupby('Holiday/Promotion')['Units Sold']
    .mean()
)

print("\n=== SALES BY DISCOUNT ===")
print(
    df_clean.groupby('Discount')['Units Sold']
    .mean()
    .sort_values(ascending=False)
)
```

```python
# STEP 9 — CORRELATION ANALYSIS

numeric_cols = [
    'Inventory Level',
    'Units Sold',
    'Units Ordered',
    'Demand Forecast',
    'Price',
    'Discount',
    'Competitor Pricing'
]

correlation = df_clean[numeric_cols].corr()

print(correlation['Units Sold'].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

# 1. Average Sales by Category
df_clean.groupby('Category')['Units Sold'].mean().sort_values().plot(
    kind='barh',
    figsize=(8,5)
)
plt.title('Average Sales by Category')
plt.xlabel('Average Units Sold')
plt.ylabel('Category')
plt.show()


# 2. Average Sales by Season
df_clean.groupby('Seasonality')['Units Sold'].mean().sort_values().plot(
    kind='bar',
    figsize=(8,5)
)
plt.title('Average Sales by Season')
plt.xlabel('Season')
plt.ylabel('Average Units Sold')
plt.show()


# 3. Actual Sales vs Demand Forecast
plt.figure(figsize=(8,5))
plt.scatter(
    df_clean['Units Sold'],
    df_clean['Demand Forecast'],
    alpha=0.3
)
plt.xlabel('Actual Units Sold')
plt.ylabel('Demand Forecast')
plt.title('Actual Sales vs Demand Forecast')
plt.show()
```

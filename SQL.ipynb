```python
result = con.execute("""
    SELECT
        Category,
        AVG("Units Sold") AS avg_units_sold
    FROM retail_inventory
    GROUP BY Category
    HAVING AVG("Units Sold") > (
        SELECT AVG("Units Sold")
        FROM retail_inventory
    )
    ORDER BY avg_units_sold DESC;
""").fetchdf()

result
```

```python
result = con.execute("""
    SELECT
        "Store ID",
        SUM("Units Sold") AS total_units_sold
    FROM retail_inventory
    GROUP BY "Store ID"
    HAVING SUM("Units Sold") > (
        SELECT AVG(total_units_sold)
        FROM (
            SELECT
                "Store ID",
                SUM("Units Sold") AS total_units_sold
            FROM retail_inventory
            GROUP BY "Store ID"
        ) AS store_totals
    )
    ORDER BY total_units_sold DESC;
""").fetchdf()

result
```

```python
result = con.execute("""
    SELECT
        Category,
        AVG("Units Sold") AS avg_units_sold
    FROM retail_inventory
    WHERE Category IN (
        SELECT Category
        FROM retail_inventory
        GROUP BY Category
        HAVING AVG("Units Sold") > (
            SELECT AVG("Units Sold")
            FROM retail_inventory
        )
    )
    GROUP BY Category
    ORDER BY avg_units_sold DESC;
""").fetchdf()

result
```

```python
result = con.execute("""
    SELECT
        "Store ID",
        SUM("Units Sold") AS total_units_sold
    FROM retail_inventory
    GROUP BY "Store ID"
    HAVING SUM("Units Sold") = (
        SELECT MAX(total_units_sold)
        FROM (
            SELECT
                "Store ID",
                SUM("Units Sold") AS total_units_sold
            FROM retail_inventory
            GROUP BY "Store ID"
        ) AS store_totals
    );
""").fetchdf()

result
```
